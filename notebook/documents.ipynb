```python
##3 create a simple a txt file
import os
os.makedirs("../data/text_files",exist_ok=True)
```

```python
sample_texts={
    "../data/text_files/rag_intro.txt":"""Retrieval-Augmented Generation (RAG) 
    is an artificial intelligence technique that improves large language model (LLM) responses by fetching relevant data from an external knowledge base before generating an answer
    How RAG Works:
    Ingestion: Documents or data sources are broken into smaller pieces called chunks and converted into numerical vectors using embedding models.
    Retrieval: A user query is turned into a vector, and a Pinecone or similar vector database performs a semantic similarity search to find the most relevant data chunks.
    Augmentation: The system combines the user query with the retrieved context into a single enhanced prompt.
    Generation: The LLM uses this enriched context to write an accurate, grounded response
    Why Use RAG?
    Reduces Hallucinations: Grounds AI answers in verified, real-world data instead of relying solely on memorized parameters.
    No Retraining Needed: Connects private company documents, policies, or live databases to public models cheaply and quickly.
    Up-to-Date Context: Keeps output fresh by referencing frequently updated external files
    """
}

for filepath,content in sample_texts.items():
    with open(filepath,'w',encoding="utf-8") as f:
        f.write(content)

print("wrote the file")
```

```python
sample_texts={
    "../data/text_files/langchain_intro.txt":"""LangChain is an open-source framework designed to simplify building applications with
     large language models (LLMs) by connecting them to external data, memory, and tools
    Core Problems LangChain SolvesNo memory: Plain LLMs treat every prompt as a new interaction.
     LangChain adds persistent memory to track conversation history.
     No external data: Models lack access to private documents or live info.
      LangChain connects models to vector databases and files via Retrieval-Augmented Generation (RAG).
      No real-world actions: Standard models only output text. 
      LangChain allows models to use external APIs, web search, and tools.
    """
}
for filepath,content in sample_texts.items():
    with open(filepath,'w',encoding="utf-8") as f:
        f.write(content)

print("wrote the file")
```

```python
from langchain_community.document_loaders import DirectoryLoader


loader = DirectoryLoader(
    "../data/text_files",
    glob="**/*.txt",            #Pattern to match files    
    loader_cls=TextLoader,     
    loader_kwargs={'encoding':"utf-8"}       ,
    show_progress=False  
)
docs = loader.load()
print(docs)
```
